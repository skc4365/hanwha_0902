```python
from langchain_openai import ChatOpenAI
from langchain_teddynote.messages import stream_response
```

```python
llm = ChatOpenAI(
    temperature=0.1,
    model="gpt-4.1-nano"
)
```

```python
req = "대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!"
res = llm.stream(req)

```

```python
for token in res:
    print(token.content, end="", flush=True)
```

```python
req2 = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")
```

```python
final_req2 = ""
for token in req2:
    final_req2 += token.content
```
