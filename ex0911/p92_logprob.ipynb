```python
from langchain_openai import ChatOpenAI
```

```python
llm_logprob = ChatOpenAI(temperature=0.1, max_completion_tokens=2048,model="gpt-4o-mini").bind(logprobs=True)
```

```python
print(llm_logprob)
```

```python
req = "대한민국의 수도는 어디인가요?"
res = llm_logprob.invoke(req)
res.response_metadata
```
